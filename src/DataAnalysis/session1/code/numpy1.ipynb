```python
def array_reshape(arr, row = -1, col = -1):
    """
    Reshape a NumPy array to a new shape.

    Parameters:
    arr (np.ndarray): The input array to reshape.
    row, col (tuple): The desired shape for the output array.

    Returns:
    np.ndarray: The reshaped array.
    """
    try:
        arr = np.array(arr)
        return arr.reshape((row, col))
    except Exception as e:
        raise ValueError (f"Error reshaping array: {e}")    

vec2 = np.array([1, 2, 3, 4, 5, 6])
mat2 = array_reshape(vec2, 3, 2)
print(mat2)
```
