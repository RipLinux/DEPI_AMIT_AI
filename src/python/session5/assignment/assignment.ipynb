```python
class Task1:
    """
    Task1 class handles validating an email address, extracting its username
    and domain, and checking the type of domain ending.

    Attributes:
        email (str): the email address to process
    """

    def __init__(self, email):
        self.email = email

    def is_valid(self):
        """
        Checks if the email has exactly one "@" and at least one "." after it.

        Returns:
            bool: True if valid, False otherwise
        """
        if self.email.count("@") != 1:
            return False
        at_index = self.email.index("@")
        after_at = self.email[at_index + 1:]
        return "." in after_at

    def get_username(self):
        """
        Extracts the part of the email before "@".

        Returns:
            str: the username, or "Invalid email" if not valid
        """
        if not self.is_valid():
            return "Invalid email"
        return self.email.split("@")[0]

    def get_domain(self):
        """
        Extracts the part between "@" and the last ".".

        Returns:
            str: the domain, or "Invalid email" if not valid
        """
        if not self.is_valid():
            return "Invalid email"
        at_index = self.email.index("@")
        last_dot_index = self.email.rindex(".")
        return self.email[at_index + 1:last_dot_index]

    def check_domain_type(self):
        """
        Checks whether the email ends with ".com", ".edu", or something else.

        Returns:
            str: "Commercial Domain", "Educational Domain", or "Other Domain"
        """
        if not self.is_valid():
            return "Invalid email"
        if self.email.endswith(".com"):
            return "Commercial Domain"
        elif self.email.endswith(".edu"):
            return "Educational Domain"
        else:
            return "Other Domain"


task1 = Task1("Amit_ml@gmail.edu")
print(task1.get_username())
print(task1.get_domain())
print(task1.check_domain_type())
```

```python
class Task2:
    """
    Task2 class decodes an encoded message by removing symbols and digits,
    reversing the first word, and shifting vowels backward in the second word.

    Attributes:
        message (str): the encoded message
    """

    def __init__(self, message):
        self.message = message
        self.vowel_map = {"A": "U", "E": "A", "I": "E", "O": "I", "U": "O"}

    def get_core(self):
        """
        Removes all symbols and digits, keeping only letters and spaces.

        Returns:
            str: the core part of the message
        """
        core = ""
        for ch in self.message:
            if ch.isalpha() or ch == " ":
                core += ch
        return core.strip()

    def decode(self):
        """
        Decodes the message by reversing the first word and shifting vowels
        in the second word.

        Returns:
            str: the decoded message
        """
        core = self.get_core()
        first_word, second_word = core.split(" ")
        reversed_first = first_word[::-1]

        new_second = ""
        for ch in second_word:
            if ch in self.vowel_map:
                new_second += self.vowel_map[ch]
            else:
                new_second += ch

        return reversed_first + " " + new_second


task2 = Task2("###!!@mocleW EPGTQ!!!6789")
print(task2.decode())
```

```python
class Task5:
    """
    Task5 class prints the multiplication table of a given number.

    Attributes:
        number (int): the number to build the table for
    """

    def __init__(self, number):
        self.number = number

    def print_table(self):
        """
        Prints the multiplication table of the number from 1 to 10.

        Returns:
            None
        """
        for i in range(1, 11):
            print(f"{self.number} x {i} = {self.number * i}")


task5 = Task5(7)
task5.print_table()
```

```python
class Task6:
    """
    Task6 class finds all twin primes less than 1000. Twin primes are two
    consecutive odd numbers that are both prime.

    Attributes:
        limit (int): the upper limit to search under
    """

    def __init__(self, limit=1000):
        self.limit = limit

    def is_prime(self, n):
        """
        Checks if a number is prime.

        Returns:
            bool: True if prime, False otherwise
        """
        if n < 2:
            return False
        for i in range(2, int(n ** 0.5) + 1):
            if n % i == 0:
                return False
        return True

    def find_twin_primes(self):
        """
        Finds all twin prime pairs below the limit.

        Returns:
            list: list of tuples, each tuple is a twin prime pair
        """
        pairs = []
        for n in range(2, self.limit - 2):
            if self.is_prime(n) and self.is_prime(n + 2):
                pairs.append((n, n + 2))
        return pairs


task6 = Task6(1000)
print(task6.find_twin_primes())
```

```python
class Task7:
    """
    Task7 class finds the prime factors of a number, including repeated
    factors.

    Attributes:
        number (int): the number to factorize
    """

    def __init__(self, number):
        self.number = number

    def get_prime_factors(self):
        """
        Finds all prime factors of the number, with repetition.

        Returns:
            list: list of prime factors
        """
        n = self.number
        factors = []
        divisor = 2
        while n > 1:
            while n % divisor == 0:
                factors.append(divisor)
                n = n // divisor
            divisor += 1
        return factors


task7 = Task7(56)
print(task7.get_prime_factors())
```

```python
class Task8:
    """
    Task8 class converts a decimal number to its binary representation
    without using the built-in bin() function.

    Attributes:
        number (int): the decimal number to convert
    """

    def __init__(self, number):
        self.number = number

    def to_binary(self):
        """
        Converts the decimal number to binary using repeated division by 2.

        Returns:
            str: the binary representation as a string
        """
        if self.number == 0:
            return "0"

        n = self.number
        binary_digits = ""
        while n > 0:
            remainder = n % 2
            binary_digits = str(remainder) + binary_digits
            n = n // 2
        return binary_digits


task8 = Task8(19)
print(task8.to_binary())
```

```python
class Task9:
    """
    Task9 class finds all perfect numbers within a given range. A number is
    perfect if the sum of its proper divisors equals the number itself.

    Attributes:
        start (int): the start of the range
        end (int): the end of the range
    """

    def __init__(self, start, end):
        self.start = start
        self.end = end

    def get_proper_divisors_sum(self, number):
        """
        Calculates the sum of proper divisors of a number.

        Returns:
            int: sum of proper divisors
        """
        total = 0
        for i in range(1, number):
            if number % i == 0:
                total += i
        return total

    def find_perfect_numbers(self):
        """
        Finds all perfect numbers in the range [start, end].

        Returns:
            list: list of perfect numbers
        """
        perfect_numbers = []
        for num in range(self.start, self.end + 1):
            if num > 1 and self.get_proper_divisors_sum(num) == num:
                perfect_numbers.append(num)
        return perfect_numbers


task9 = Task9(1, 10000)
print(task9.find_perfect_numbers())
```
