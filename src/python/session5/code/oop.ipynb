```python
class Student:
    def __init__(self, name: str = "name", age: int = 0):
        self.name = name
        self.age = age

    def display(self):
        print(f"Name: {self.name}")
        print("Age:", self.age)
```

```python
s11 = Student("Anas")
s11.display()
```

```python
from abc import ABC, abstractmethod
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius # public

    def area(self):
        return 3.14 * self.radius ** 2
class Rectangle(Shape):
    def __init__(self, width, height):
        self._width = width # protected var
        self.__height = height # private var

    def area(self):
        return self.width * self.height




```

```python
Circle1 = Circle(5)
print("Area of Circle:", Circle1.area())
```

```python
class Employee:
    def __init__(self,  name:str ,salary: int):
        self.salary = salary
        self.name = name
    def set_salary(self,obj,new_salary):
        obj.salary = new_salary
class Manager:
    def __init__(self, name:str , salary:int):
        self.name = name 
        self._salary = salary
```

```python
e1 = Employee("ahmed",5000)
m1 = Manager("anas",1000)
e1.set_salary(m1,6000)
```
