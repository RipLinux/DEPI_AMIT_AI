```python
def prime_factors(num: int):
    divisor_numbers = []      # store all divisors of the number
    prime_factors = []      # store the final prime factors
    '''
    First loop: Collect every number that divides num without remainder.
    Example: num = 56
    The list will be:
    [2, 4, 7, 8, 14, 28]
    '''
    for divisor_number in range(2, num):
        if num % divisor_number == 0:
            divisor_numbers.append(divisor_number)
    '''
    Second loop: Check every divisor.
    We use while instead of if because a prime factor
    may appear more than once.
    Example: num = 56
    First divisor = 2
    56 % 2 == 0
    Add 2 and num = 56 // 2 = 28
    Check 2 again
    28 % 2 == 0
    Add 2 and num = 28 // 2 = 14
    Check 2 again
    14 % 2 == 0
    Add 2 and num = 14 // 2 = 7
    Check 2 again
    7 % 2 != 0
    Exit while and move to the correct divisor.
    7 -> 7 % 7 == 0
    Add 7 + num = 7 // 7 = 1
    No other divisor can divide it.
    '''
    for prime_factor in divisor_numbers:
        while num % prime_factor == 0:
            prime_factors.append(prime_factor)
            num //= prime_factor

    return prime_factors

print(prime_factors(56))
```
