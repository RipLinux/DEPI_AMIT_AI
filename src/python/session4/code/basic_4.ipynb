```python
def generate_password(length:int):
    char = "abcdefghopqlmnzxyzABCDEFGOPZLEKPEIOPK123456789!@#$%^&*"
    password = ""
    for _ in range(length):
        password += random.choice(char)
    return password
generate_password(12)
```

```python
import string
import secrets
def generate_password_secure(length:int =12):
    if length <=0:
        raise ValueError ("Password length must be more than zero")
    char = ( 
        string.ascii_letters +
        string.digits +
        string.punctuation
    )
    return "".join(secrets.choice (char) for _ in range(length))
generate_password_secure(12)
```

```python
def count_words(words: list):
    unique_words = set(words)
    counts = {word: 0 for word in unique_words}

    for i in unique_words:
        for j in words:
            if i == j:
                counts[i] += 1

    return counts


words = ["lol", "anas", "lol", "hi", "lol", "hi"]


print(count_words(words))
```

```python
import random

responses = {
    "hello": [
        "Hello!",
        "Hi there!",
        "Greetings!"
    ],
    "hi": [
        "Hi!",
        "Hello!",
        "Hey!"
    ],
    "how are you": [
        "I'm doing great!",
        "I'm fine, thanks!",
        "I'm awesome! How about you?"
    ],
    "bye": [
        "Goodbye!",
        "See you later!",
        "Take care!"
    ],
    "thanks": [
        "You're welcome!",
        "No problem!",
        "Happy to help!"
    ],
    "name": [
        "I'm a simple chatbot.",
        "You can call me ChatBot.",
        "I'm your Python assistant."
    ]
}

def chatbot(message):
    message = message.lower().strip()

    if message in responses:
        return random.choice(responses[message])

    return random.choice([
        "Sorry, I don't understand.",
        "Can you say that another way?",
        "I'm not sure what you mean."
    ])


while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Bot: Goodbye!")
        break

    print("Bot:", chatbot(user))
```
