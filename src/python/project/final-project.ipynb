### Task 1 - Bank Account

```python
class BankAccount:
    """
    Class representing a simple bank account.
    Converted from function-based deposit/withdraw/check_balance functions.
    """

    def __init__(self, initial_balance=0):
        """
        Initializes the account with a starting balance.
        """
        self.balance = initial_balance

    def deposit(self, amount):
        """
        Adds a positive amount to the balance.
        Returns: None
        """
        if amount <= 0:
            print("Error: Deposit amount must be greater than zero.")
        else:
            self.balance += amount
            print(f"Deposited {amount}. New balance: {self.balance}")

    def withdraw(self, amount):
        """
        Subtracts a positive amount from the balance if funds are sufficient.
        Returns: None
        """
        if amount <= 0:
            print("Error: Withdraw amount must be greater than zero.")
        elif amount > self.balance:
            print("Error: Insufficient balance.")
        else:
            self.balance -= amount
            print(f"Withdrew {amount}. New balance: {self.balance}")

    def check_balance(self):
        """
        Prints the current balance.
        Returns: the balance (int/float)
        """
        print(f"Current balance: {self.balance}")
        return self.balance


if __name__ == "__main__":
    account = BankAccount(100)
    account.deposit(50)
    account.withdraw(30)
    account.check_balance()
```

### Task 2 - Calculator

```python
class Calculator:
    """
    Class implementing a simple calculator with a menu-driven interface.
    """

    def __init__(self):
        """
        No special initialization needed for this calculator.
        """
        pass

    def add(self, a, b):
        """
        Returns the sum of a and b.
        """
        return a + b

    def subtract(self, a, b):
        """
        Returns the result of a minus b.
        """
        return a - b

    def multiply(self, a, b):
        """
        Returns the product of a and b.
        """
        return a * b

    def divide(self, a, b):
        """
        Returns a divided by b. Prevents division by zero.
        Returns: result (float) or None if b is zero.
        """
        if b == 0:
            print("Error: Cannot divide by zero.")
            return None
        return a / b

    def calculator(self):
        """
        Displays the menu, takes user input, and calls the correct method.
        Returns: None
        """
        print("Welcome to the calculator!")
        print("1: Add")
        print("2: Subtract")
        print("3: Multiply")
        print("4: Divide")

        choice = input("Choose an option (1-4): ")

        try:
            num1 = float(input("Enter first number: "))
            num2 = float(input("Enter second number: "))
        except ValueError:
            print("Error: Please enter valid numbers.")
            return

        if choice == "1":
            print(f"Result: {self.add(num1, num2)}")
        elif choice == "2":
            print(f"Result: {self.subtract(num1, num2)}")
        elif choice == "3":
            print(f"Result: {self.multiply(num1, num2)}")
        elif choice == "4":
            result = self.divide(num1, num2)
            if result is not None:
                print(f"Result: {result}")
        else:
            print("Error: Invalid choice. Please select 1-4.")


if __name__ == "__main__":
    calculator = Calculator()
    calculator.calculator()
```

### Task 3 - Abstract Animal Class

```python
from abc import ABC, abstractmethod


class Animal(ABC):
    """
    Abstract base class representing a generic animal.
    Cannot be instantiated directly.
    """

    def __init__(self, name):
        """
        Initializes the animal with a name.
        """
        self.name = name

    @abstractmethod
    def make_sound(self):
        """
        Abstract method that must be implemented by subclasses.
        Returns: the sound the animal makes (str)
        """
        pass

    def describe(self):
        """
        Prints a description of the animal, including its sound.
        Returns: None
        """
        print(f"{self.name} says: {self.make_sound()}")


class Dog(Animal):
    """
    Represents a dog. Inherits from Animal.
    """

    def make_sound(self):
        """
        Returns the sound a dog makes.
        """
        return "Woof!"


class Cat(Animal):
    """
    Represents a cat. Inherits from Animal.
    """

    def make_sound(self):
        """
        Returns the sound a cat makes.
        """
        return "Meow!"


class Cow(Animal):
    """
    Represents a cow. Inherits from Animal.
    """

    def make_sound(self):
        """
        Returns the sound a cow makes.
        """
        return "Moo!"


if __name__ == "__main__":
    dog = Dog("Rex")
    cat = Cat("Whiskers")
    cow = Cow("Bessie")

    dog.make_sound()
    dog.describe()

    cat.make_sound()
    cat.describe()

    cow.make_sound()
    cow.describe()
```

### Task 4 - Text File Reader

```python
class TextFileReader:
    """
    Class for reading and analyzing a text file.
    """

    def __init__(self, file_path):
        """
        Initializes the reader with a file path and empty content.
        """
        self.file_path = file_path
        self.content = ""

    def read_file(self):
        """
        Opens the file and stores its content in self.content.
        Returns: None
        """
        try:
            with open(self.file_path, "r", encoding="utf-8") as file:
                self.content = file.read()
            print("File read successfully.")
        except FileNotFoundError:
            print(f"Error: File '{self.file_path}' was not found.")
        except IOError:
            print(f"Error: A problem occurred while reading '{self.file_path}'.")

    def count_lines(self):
        """
        Returns the number of lines in the file content.
        """
        if not self.content:
            return 0
        return len(self.content.splitlines())

    def count_words(self):
        """
        Returns the number of words in the file content.
        """
        return len(self.content.split())

    def count_characters(self):
        """
        Returns the number of characters in the file content.
        """
        return len(self.content)

    def display_content(self):
        """
        Prints the content of the file.
        Returns: None
        """
        print(self.content)


if __name__ == "__main__":
    reader = TextFileReader("test.txt")
    reader.read_file()
    reader.display_content()
    print(f"Lines: {reader.count_lines()}")
    print(f"Words: {reader.count_words()}")
    print(f"Characters: {reader.count_characters()}")
```

### Task 5 - Unique Words and Frequency

```python
def count_word_frequency(words):
    """
    Counts how many times each word appears in a list.
    Returns: a dictionary mapping each word to its frequency.
    """
    unique_words = set(words)  # extract unique words
    frequency = {}

    for word in unique_words:
        count = 0
        for w in words:
            if w == word:
                count += 1
        frequency[word] = count

    return frequency


if __name__ == "__main__":
    words = [
        "Welcome",
        "Ali",
        "Hi",
        "Ali",
        "No",
        "Hi",
        "No",
        "Ali",
        "No",
        "Ali"
    ]

    result = count_word_frequency(words)
    print(result)
```

### Task 6 - Read File , User Extractor

```python
def read_txt_file(file_path):
    """
    Opens a text file and returns its content as a string.
    Returns: file content (str), or None if an error occurs.
    """
    try:
        with open(file_path, "r", encoding="utf-8") as file:
            content = file.read()
        return content
    except FileNotFoundError:
        print(f"Error: File '{file_path}' was not found.")
        return None
    except IOError:
        print(f"Error: A problem occurred while reading '{file_path}'.")
        return None


class UserExtractor:
    """
    Class for extracting unique usernames from a 'username:password' file.
    """

    def __init__(self, file_path):
        """
        Initializes the extractor with a file path and an empty usernames dictionary.
        """
        self.file_path = file_path
        self.usernames = {}

    def extract_usernames(self):
        """
        Reads the file using read_txt_file() and extracts unique usernames.
        Returns: dictionary of unique usernames
        """
        content = read_txt_file(self.file_path)

        if content is None:
            print("Could not extract usernames due to a file error.")
            return self.usernames

        lines = content.splitlines()

        for line in lines:
            if ":" in line:
                username = line.split(":")[0].strip()
                if username not in self.usernames:
                    self.usernames[username] = True  # mark as seen

        return self.usernames


if __name__ == "__main__":
    extractor = UserExtractor("users.txt")
    result = extractor.extract_usernames()
    print(list(result.keys()))
```
